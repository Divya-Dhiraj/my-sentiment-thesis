```python
import pandas as pd

# Load your real data using the full path
# low_memory=False is added to handle the DtypeWarning
full_file_path = r"C:\projects\my-local-thesis\data\processed_data.csv"
df = pd.read_csv(full_file_path, sep='\t', low_memory=False)

# Convert date column to datetime objects for proper grouping
df['ship_day'] = pd.to_datetime(df['ship_day'])

# 1. Split the DataFrame into orders and returns based on your business rule
orders_df = df[df['concession_reason'].isnull()].copy()
returns_df = df[df['concession_reason'].notnull()].copy()

# 2. Aggregate the orders data to get sales figures
weekly_sales = orders_df.groupby(
    [pd.Grouper(key='ship_day', freq='W-MON'), 'asin']
).agg(
    total_units_sold=('shipped_units', 'sum'),
    average_selling_price=('our_price', 'mean')
).reset_index()

# 3. Aggregate the returns data to get concession figures
weekly_returns = returns_df.groupby(
    [pd.Grouper(key='ship_day', freq='W-MON'), 'asin']
).agg(
    total_units_conceded=('total_units_conceded', 'sum')
).reset_index()

# 4. Merge the two summaries into one comprehensive weekly report
weekly_performance_df = pd.merge(
    weekly_sales,
    weekly_returns,
    on=['ship_day', 'asin'],
    how='outer' # 'outer' merge keeps all records from both tables
)

# 5. Clean up the final DataFrame by filling empty values with 0
weekly_performance_df.fillna(0, inplace=True)

print("✅ Successfully created a combined weekly performance summary:")
print(weekly_performance_df.head())
```

```python
import re
import pandas as pd
from sqlalchemy import text
from thefuzz import process
from functools import lru_cache

# We assume 'engine' is already loaded in your notebook from a previous cell.
# If not, you would add the path-fixing code and `from src.database import engine` here.

# --- Helper Function (Copied from custom_tools.py) ---
# We need this because the main function calls it.
def get_dates_from_query(query: str) -> dict:
    # For this test, we'll use a simplified version.
    # A full version would query the DB to get the min/max dates.
    # This is sufficient for our current test.
    min_db_date, max_db_date = "2021-01-01", "2025-12-31" # Assuming a wide range for testing
    
    if "december 2024" in query.lower():
        return {"start_date": "2024-12-01", "end_date": "2024-12-31"}
    # Add more simple date logic here if needed for other tests
    
    # Default fallback
    return {"start_date": min_db_date, "end_date": max_db_date}


# --- Main Function (This is our new, upgraded version) ---
# You can modify this function as much as you want.
def financial_data_tool(query: str) -> str:
    """
    Returns weekly performance data for a product within a specified date range.
    It first searches for a direct ASIN match in the query. If none is found,
    it uses fuzzy matching to find products by name.
    """
    print(f"--- PerformanceAgent: Received query: '{query}' ---")
    try:
        product_id_found = None
        product_name_found = None
        
        # Step 1: Look for a direct ASIN match
        asin_match = re.search(r'\bB[A-Z0-9]{9}\b', query)
        if asin_match:
            product_id_found = asin_match.group(0)
            print(f"--- PerformanceAgent: Found direct ASIN match: {product_id_found} ---")
            with engine.connect() as conn:
                result = conn.execute(text("SELECT product_name FROM products WHERE asin = :asin"), {"asin": product_id_found}).fetchone()
                if result:
                    product_name_found = result[0]
                else:
                    return f"Found ASIN {product_id_found} in query, but this product does not exist in the database."

        # Step 2: Fallback to fuzzy name matching
        if not product_id_found:
            print("--- PerformanceAgent: No ASIN found, falling back to fuzzy name search. ---")
            products_df = pd.read_sql_query("SELECT asin, product_name FROM products;", engine)
            product_choices = products_df['product_name'].tolist()
            best_match = process.extractOne(query, product_choices)
            if best_match and best_match[1] >= 80:
                product_name_found = best_match[0]
                product_id_found = products_df[products_df['product_name'] == product_name_found]['asin'].iloc[0]
            else:
                return "Sorry, I couldn't find a matching product by name."

        if not product_id_found:
             return "Sorry, I couldn't identify a product from your query."

        # Step 3: Get dates and query the data
        dates = get_dates_from_query(query)
        start_date, end_date = dates['start_date'], dates['end_date']
        print(f"--- PerformanceAgent: Matched to '{product_name_found}' (ASIN: {product_id_found}), Date Range: {start_date} to {end_date} ---")

        with engine.connect() as conn:
            sql_query = """
                SELECT week_start_date, total_units_sold, average_selling_price, num_reviews_received 
                FROM weekly_performance 
                WHERE asin = :asin AND week_start_date BETWEEN :start_date AND :end_date
                ORDER BY week_start_date;
            """
            params = {'asin': product_id_found, 'start_date': start_date, 'end_date': end_date}
            data = pd.read_sql_query(text(sql_query), conn, params=params)
        
        if data.empty:
            return f"No performance data found for {product_name_found} in the period ({start_date} to {end_date})."
        
        return f"Product Name: {product_name_found}\nPerformance data from {start_date} to {end_date}:\n{data.to_markdown(index=False)}"

    except Exception as e:
        import traceback
        traceback.print_exc()
        return f"Error fetching performance data: {e}"


# --- Test Call ---
# Now we call the function we just defined in this cell
test_query = "What were the sales for ASIN B08NJNKG7R in December 2024?"
result = financial_data_tool(test_query)

print("\n--- Final Result from the Tool ---")
print(result)
```

```python
import pandas as pd
import os
import json
from enum import Enum
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field
from typing import List, Optional
from tqdm import tqdm
import sys
from dotenv import load_dotenv

# --- SETUP ---
dotenv_path = os.path.join('..', '.env')
load_dotenv(dotenv_path=dotenv_path)
print("--- Loaded configuration from .env file ---")
print("\n--- Starting Robust & Fast Data Enrichment Process ---")

# File paths
source_file = os.path.join('..', 'data', 'refined_data.csv')
partial_output_file = os.path.join('..', 'data', 'enriched_data_partial.csv')
final_output_file = os.path.join('..', 'data', 'enriched_data.csv')

# --- 1. DEFINE PYDANTIC MODELS & ENUMS (Unchanged) ---
# ... (All your Enum and BaseModel classes from the previous step go here)
class MarketSegmentEnum(str, Enum): APPLIANCE_PARTS_ACCESSORIES = "Appliance Parts & Accessories"; FREESTANDING_APPLIANCES = "Freestanding Appliances"; CLEANING_AND_CARE = "Cleaning & Care"; UNKNOWN = "Unknown"
class PurchaseIntentEnum(str, Enum): REPAIR_MAINTENANCE = "Repair/Maintenance"; CONSUMABLE_REPLACEMENT = "Consumable Replacement"; NEW_APPLIANCE_PURCHASE = "New Appliance Purchase"; ACCESSORY_UPGRADE = "Accessory/Upgrade"; UNKNOWN = "Unknown"
class ReturnThemeEnum(str, Enum): COMPATIBILITY_FIT_ISSUE = "Compatibility/Fit Issue"; PRODUCT_DEFECT_OR_DAMAGE = "Product Defect or Damage"; SHIPPING_ISSUE = "Shipping Issue (Damaged, Late, Lost)"; WRONG_ITEM_ORDERED_OR_RECEIVED = "Wrong Item Ordered or Received"; NO_LONGER_NEEDED = "No Longer Needed/Wanted"; PRICE_OR_VALUE_ISSUE = "Price or Value Issue"; MISSING_PARTS = "Missing Parts or Accessories"; DESCRIPTION_MISMATCH = "Description Mismatch"; UNKNOWN_OR_NO_REASON_GIVEN = "Unknown or No Reason Given"
class SuggestedActionEnum(str, Enum): UPDATE_PRODUCT_LISTING = "Update Product Listing (Description, Images, Compatibility Info)"; REVIEW_SUPPLIER_QUALITY_CONTROL = "Review Supplier Quality Control"; IMPROVE_SHIPPING_PACKAGING = "Improve Shipping Packaging and Carrier Performance"; REVIEW_WAREHOUSE_PICKING_PROCESS = "Review Warehouse Picking & Packing Accuracy"; NO_IMMEDIATE_ACTION_NEEDED = "No Immediate Action Needed (Customer Discretionary Return)"; REVIEW_PRICING_AND_PROMOTIONS = "Review Pricing and Promotions"; VERIFY_PACKING_LIST_ACCURACY = "Verify Packing List & Component Accuracy"; AUDIT_CARRIER_PERFORMANCE = "Audit Carrier Performance and Logistics"; DATA_GAP_ANALYSIS = "Data Gap Analysis (for 'No Reason Given')"; UNKNOWN = "Unknown"
class SentimentEnum(str, Enum): NEGATIVE = "Negative"; NEUTRAL = "Neutral"; POSITIVE = "Positive"

class ProductEnrichment(BaseModel):
    key: str = Field(description="The unique key combining subcategory and product type.")
    ProductName: str; PurchaseIntent: PurchaseIntentEnum; MarketSegment: MarketSegmentEnum; PredictedDiscountImpact: str
class ReturnEnrichment(BaseModel):
    key: str = Field(description="The original concession reason."); Sentiment: SentimentEnum; ReturnTheme: ReturnThemeEnum; SuggestedAction: SuggestedActionEnum

class ProductEnrichmentList(BaseModel):
    products: List[ProductEnrichment]
class ReturnEnrichmentList(BaseModel):
    returns: List[ReturnEnrichment]

# --- 2. SETUP SPECIALIZED LLM CHAINS ---
llm = ChatOpenAI(api_key=os.environ.get("OPENAI_API_KEY"), model=os.environ.get("OPENAI_MODEL_NAME"), model_kwargs={"response_format": {"type": "json_object"}})

# --- THIS IS THE FIX: Update prompts to include the word "JSON" ---
# Chain 1: For Products
product_parser = JsonOutputParser(pydantic_object=ProductEnrichmentList)
product_prompt = ChatPromptTemplate.from_template(
    "You are a product data analyst. For each item in the input list, generate the requested enrichment attributes. Your response must be a single, valid JSON object with a 'products' key, containing a list of the enriched data.\n"
    "Input List: {product_list_json}\n"
    "JSON Format Instructions: {format_instructions}"
)
product_chain = product_prompt | llm | product_parser

# Chain 2: For Returns
return_parser = JsonOutputParser(pydantic_object=ReturnEnrichmentList)
return_prompt = ChatPromptTemplate.from_template(
    "You are a returns analyst. For each reason in the input list, generate the requested attributes. Your response must be a single, valid JSON object with a 'returns' key, containing a list of the enriched data.\n"
    "Input List: {reasons_list_json}\n"
    "JSON Format Instructions: {format_instructions}"
)
return_chain = return_prompt | llm | return_parser

# --- 3. MAIN EXECUTION SCRIPT ---
try:
    print(f"Loading data from {source_file}...")
    df = pd.read_csv(source_file, low_memory=False, keep_default_na=False)
    print(f"✅ Successfully loaded {len(df)} total rows.")
    
    if os.path.exists(partial_output_file):
        print(f"--- Found partial file at {partial_output_file}. Resuming process. ---")
        enriched_df = pd.read_csv(partial_output_file)
    else:
        enriched_df = df[['asin']].copy()

    # --- BATCH ENRICHMENT: PRODUCTS ---
    if 'ProductName' not in enriched_df.columns:
        print("\n--- Step 1: Enriching Product Data in a Single Batch ---")
        unique_products = df[['subcategory_description', 'product_type']].drop_duplicates().reset_index(drop=True)
        unique_products['key'] = unique_products['subcategory_description'].astype(str) + "||" + unique_products['product_type'].astype(str)
        product_list_json = unique_products.to_json(orient='records')
        
        print(f"Found {len(unique_products)} unique product combinations to enrich...")
        product_response = product_chain.invoke({
            "product_list_json": product_list_json,
            "format_instructions": product_parser.get_format_instructions()
        })
        enriched_products_df = pd.DataFrame(product_response['products'])
        
        enriched_products_df[['subcategory_description', 'product_type']] = enriched_products_df['key'].str.split('||', n=1, expand=True)
        enriched_df = pd.merge(df, enriched_products_df.drop('key', axis=1), on=['subcategory_description', 'product_type'], how='left')
        
        enriched_df.to_csv(partial_output_file, index=False)
        print("✅ Product enrichment complete. Checkpoint saved.")

    # --- BATCH ENRICHMENT: RETURNS ---
    if 'ReturnTheme' not in enriched_df.columns:
        print("\n--- Step 2: Enriching Return Data in a Single Batch ---")
        unique_reasons = df[df['concession_reason'].notna() & (df['concession_reason'] != '') & (df['concession_reason'] != 'N/A')]['concession_reason'].unique().tolist()
        
        print(f"Found {len(unique_reasons)} unique return reasons to enrich...")
        return_response = return_chain.invoke({
            "reasons_list_json": json.dumps([{"key": r} for r in unique_reasons]),
            "format_instructions": return_parser.get_format_instructions()
        })
        enriched_reasons_df = pd.DataFrame(return_response['returns'])
        enriched_reasons_df.rename(columns={'key': 'concession_reason'}, inplace=True)

        enriched_df = pd.merge(enriched_df, enriched_reasons_df, on='concession_reason', how='left')
        
        enriched_df.to_csv(partial_output_file, index=False)
        print("✅ Return enrichment complete. Checkpoint saved.")
        
    # --- FINALIZATION ---
    print("\nFinalizing enriched data...")
    final_df = pd.read_csv(partial_output_file)
    final_df.to_csv(final_output_file, index=False)
    if os.path.exists(partial_output_file):
        os.remove(partial_output_file)

    print("\n🎉 --- Data Enrichment Complete! --- 🎉")
    print(f"A new file 'enriched_data.csv' has been created in your 'data' directory.")

except Exception as e:
    import traceback
    traceback.print_exc()
```

```python
import pandas as pd
import os
import json
from enum import Enum
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field
from typing import List, Optional
from tqdm import tqdm
import sys
from dotenv import load_dotenv

# Load environment variables from the .env file in the project root
dotenv_path = os.path.join('..', '.env')
load_dotenv(dotenv_path=dotenv_path)
print("--- Loaded configuration from .env file ---")

print("\n--- Starting Final, Separated Data Enrichment Process ---")

# --- FILE PATHS ---
source_file = os.path.join('..', 'data', 'refined_data.csv')
output_file = os.path.join('..', 'data', 'enriched_data.csv')

# --- 1. DEFINE DETAILED ENUMS AND PYDANTIC MODELS ---

class MarketSegmentEnum(str, Enum):
    APPLIANCE_PARTS_ACCESSORIES = "Appliance Parts & Accessories"
    FREESTANDING_APPLIANCES = "Freestanding Appliances"
    CLEANING_AND_CARE = "Cleaning & Care"
    UNKNOWN = "Unknown"

class PurchaseIntentEnum(str, Enum):
    REPAIR_MAINTENANCE = "Repair/Maintenance"
    CONSUMABLE_REPLACEMENT = "Consumable Replacement"
    NEW_APPLIANCE_PURCHASE = "New Appliance Purchase"
    ACCESSORY_UPGRADE = "Accessory/Upgrade"
    UNKNOWN = "Unknown"

class ReturnThemeEnum(str, Enum):
    COMPATIBILITY_FIT_ISSUE = "Compatibility/Fit Issue"
    PRODUCT_DEFECT_OR_DAMAGE = "Product Defect or Damage"
    SHIPPING_ISSUE = "Shipping Issue (Damaged, Late, Lost)"
    WRONG_ITEM_ORDERED_OR_RECEIVED = "Wrong Item Ordered or Received"
    NO_LONGER_NEEDED = "No Longer Needed/Wanted"
    PRICE_OR_VALUE_ISSUE = "Price or Value Issue"
    MISSING_PARTS = "Missing Parts or Accessories"
    DESCRIPTION_MISMATCH = "Description Mismatch"
    UNKNOWN_OR_NO_REASON_GIVEN = "Unknown or No Reason Given"

class SuggestedActionEnum(str, Enum):
    UPDATE_PRODUCT_LISTING = "Update Product Listing (Description, Images, Compatibility Info)"
    REVIEW_SUPPLIER_QUALITY_CONTROL = "Review Supplier Quality Control"
    IMPROVE_SHIPPING_PACKAGING = "Improve Shipping Packaging and Carrier Performance"
    REVIEW_WAREHOUSE_PICKING_PROCESS = "Review Warehouse Picking & Packing Accuracy"
    NO_IMMEDIATE_ACTION_NEEDED = "No Immediate Action Needed (Customer Discretionary Return)"
    REVIEW_PRICING_AND_PROMOTIONS = "Review Pricing and Promotions"
    VERIFY_PACKING_LIST_ACCURACY = "Verify Packing List & Component Accuracy"
    AUDIT_CARRIER_PERFORMANCE = "Audit Carrier Performance and Logistics"
    DATA_GAP_ANALYSIS = "Data Gap Analysis (for 'No Reason Given')"
    UNKNOWN = "Unknown"

class SentimentEnum(str, Enum):
    NEGATIVE = "Negative"
    NEUTRAL = "Neutral"
    POSITIVE = "Positive"

# Pydantic model for the output of the product enrichment chain
class ProductEnrichment(BaseModel):
    ProductName: str = Field(description="A concise, user-friendly product name.")
    PurchaseIntent: PurchaseIntentEnum
    MarketSegment: MarketSegmentEnum
    PredictedDiscountImpact: str = Field(description="Predicted sales impact of a 10% discount ('High', 'Medium', 'Low').")

# Pydantic model for the output of the return enrichment chain
class ReturnEnrichment(BaseModel):
    Sentiment: SentimentEnum
    ReturnTheme: ReturnThemeEnum
    SuggestedAction: SuggestedActionEnum

# --- 2. SETUP SPECIALIZED LLM CHAINS ---

llm = ChatOpenAI(
    api_key=os.environ.get("OPENAI_API_KEY"),
    model=os.environ.get("OPENAI_MODEL_NAME"),
    model_kwargs={"response_format": {"type": "json_object"}}
)

# Chain 1: For Products/Sales
product_parser = JsonOutputParser(pydantic_object=ProductEnrichment)
product_prompt = ChatPromptTemplate.from_template(
    "You are a product data analyst. Based on the product details, generate the requested attributes in a valid JSON format. Choose the BEST fitting option for each field.\n"
    "Product Details:\n- Subcategory: {subcategory_description}\n- Type: {product_type}\n"
    "Available Purchase Intents: {purchase_intents}\n"
    "Available Market Segments: {market_segments}\n"
    "JSON Format Instructions: {format_instructions}"
)
product_enrichment_chain = product_prompt.partial(
    purchase_intents=", ".join([f"'{e.value}'" for e in PurchaseIntentEnum]),
    market_segments=", ".join([f"'{e.value}'" for e in MarketSegmentEnum]),
    format_instructions=product_parser.get_format_instructions()
) | llm | product_parser

# Chain 2: For Returns
return_parser = JsonOutputParser(pydantic_object=ReturnEnrichment)
return_prompt = ChatPromptTemplate.from_template(
    "You are a returns analyst. Based on the customer return reason, generate the requested attributes in a valid JSON format. Choose the BEST fitting option for each field.\n"
    "Return Reason: {concession_reason}\n"
    "Available Sentiments: 'Negative', 'Neutral', 'Positive'\n"
    "Available Return Themes: {return_themes}\n"
    "Available Suggested Actions: {suggested_actions}\n"
    "JSON Format Instructions: {format_instructions}"
)
return_enrichment_chain = return_prompt.partial(
    return_themes=", ".join([f"'{e.value}'" for e in ReturnThemeEnum]),
    suggested_actions=", ".join([f"'{e.value}'" for e in SuggestedActionEnum]),
    format_instructions=return_parser.get_format_instructions()
) | llm | return_parser


# --- 3. LOAD AND SPLIT THE DATA ---
try:
    print(f"Loading data from {source_file}...")
    df = pd.read_csv(source_file, low_memory=False, keep_default_na=False)
    print(f"✅ Successfully loaded {len(df)} total rows.")

    is_return = (df['total_units_conceded'] > 0) & (df['concession_reason'].notna()) & (df['concession_reason'] != '') & (df['concession_reason'] != 'N/A')
    returns_df = df[is_return].copy()
    sales_df = df[~is_return].copy()
    print(f"Separated data into {len(sales_df)} sales records and {len(returns_df)} return records.")

    # --- 4. ENRICH SALES AND RETURNS SEPARATELY ---
    # Use a smaller sample for testing. Remove .head() for the full run.
    sales_sample = sales_df.head(5) 
    returns_sample = returns_df.head(5)

    enriched_sales_list = []
    print(f"\nEnriching {len(sales_sample)} sales records...")
    for index, row in tqdm(sales_sample.iterrows(), total=sales_sample.shape[0]):
        try:
            input_data = {"subcategory_description": row.get('subcategory_description', ''), "product_type": row.get('product_type', '')}
            enriched_sales_list.append(product_enrichment_chain.invoke(input_data))
        except Exception as e:
            print(f"Error on sales row {index}: {e}")
            enriched_sales_list.append({})

    enriched_returns_list = []
    print(f"\nEnriching {len(returns_sample)} return records...")
    for index, row in tqdm(returns_sample.iterrows(), total=returns_sample.shape[0]):
        try:
            product_input = {"subcategory_description": row.get('subcategory_description', ''), "product_type": row.get('product_type', '')}
            product_details = product_enrichment_chain.invoke(product_input)
            
            return_input = {"concession_reason": row.get('concession_reason', '')}
            return_details = return_enrichment_chain.invoke(return_input)
            
            product_details.update(return_details)
            enriched_returns_list.append(product_details)
        except Exception as e:
            print(f"Error on return row {index}: {e}")
            enriched_returns_list.append({})

    # --- 5. COMBINE AND SAVE ---
    print("\nCombining original data with new enriched columns...")
    
    enriched_sales_df = pd.DataFrame(enriched_sales_list, index=sales_sample.index)
    enriched_returns_df = pd.DataFrame(enriched_returns_list, index=returns_sample.index)
    
    # Update the sample DataFrames with the new data
    sales_sample = sales_sample.assign(**enriched_sales_df)
    returns_sample = returns_sample.assign(**enriched_returns_df)

    # Combine the enriched samples back together
    final_df = pd.concat([sales_sample, returns_sample]).sort_index()

    print(f"Saving the final enriched sample data to {output_file}...")
    final_df.to_csv(output_file, index=False)

    print("\n🎉 --- Data Enrichment Complete! --- 🎉")
    print(f"A new file 'enriched_data.csv' with {len(final_df.columns)} columns has been created in your 'data' directory.")

except FileNotFoundError:
    print(f"❌ ERROR: The file was not found at '{source_file}'.")
except Exception as e:
    import traceback
    traceback.print_exc()
```
